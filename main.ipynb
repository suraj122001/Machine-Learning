```python
import ast
#ast.literal_eval() used to convert strings into list
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i["name"])
    return L

movies["genres"]=movies["genres"].apply(convert)

```

```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i["name"])
    return L

movies["keywords"]=movies["keywords"].apply(convert)
```

```python
def convert3(obj):
    L=[]
    j=0
    for i in ast.literal_eval(obj):
        if j!=3:
            L.append(i["name"])
            j+=1
        else:
            break
    return L

movies["cast"]=movies["cast"].apply(convert3)
```

```python
def hunt(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i["job"]=="Director":
            L.append(i["name"])
            break
    return L

movies["crew"]=movies["crew"].apply(hunt)
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_movies["tags"]=new_movies["tags"].apply(stem)
```

```python
similarity=cosine_similarity(vectors)
```

```python
def recommend(movie): 
    movie_index=new_movies[new_movies["title"] == movie].index[0]
    distance=similarity[movie_index]
    movie_list=sorted(list(enumerate(distance)),reverse=True,key=lambda x : x[1])[1:6]
    for i in movie_list:
        print(new_movies.iloc[i[0]]["title"])  #here we get the names using just the id
    # v=int(x)
    
```
